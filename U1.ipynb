### read and write (r+)

```python
f=open('H1.txt','r+')
print(f.read())
f.write('New data\n')
f.write('My Data')
f.close()
```

```python
f=open('H1.txt','r+')
f.write('New data\n')
f.write('My Data')
f.close()
```

#### f.seek()
#### f.tell()

```python
f=open('H1.txt','r+')
print(f.read(5))
print(f.seek(5))
f.write('New data\n')
f.write('My Data')
f.close()
```

### write mode(w)

```python
f=open('H1.txt','w')
f.write('this is test message')
f.close()
```

```python
f=open('H1.txt','w+')
f.write('this is test message')
print(f.tell())
print(f.read())
f.close()
```

```python
f=open('H1.txt','w+')
f.write('this is test message')
print(f.tell())
f.seek(0)
print(f.read())
f.close()
```

```python
f=open('Demo.txt','w+')
f.write('this is test message')
f.seek(0)
print(f.read())
f.close()
```

### Reading the files
1) read()

```python
f=open('H1.txt','r')
print(f.read(5))
f.close()
```

```python
f=open('H1.txt','r')
print(f.read(5))
print(f.read(10))
print(f.read(7))
print(f.read())
f.close()
```

```python
f=open('H1.txt','r')
print(f.read(5))
print(f.read())
print(f.read(7))
print(f.read())
f.close()
```

2) read line()

```python
f=open('H1.txt','r')
f.readline()
f.readline()
f.readline()
print(f.readline())
f.close()
```

```python
f=open('H1.txt','r')
f.readline(2)
f.readline(3)
print(f.readline())
f.close()
```

```python
f=open('H1.txt','r')
print(f.readline(2))
print(f.readline(3))
print(f.readline())
f.close()
```

```python
f=open('H1.txt','r')
print(f.readline(20))
print(f.readline(3))
print(f.readline())
f.close()
```

3) readlines()

```python
f=open('H1.txt','r')
print(f.readlines())
f.close()
```

```python
f=open('H1.txt','r')
x=f.readlines()
print(x[3])
f.close()
```
