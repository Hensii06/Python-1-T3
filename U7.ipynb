### deleting directory or files
- os.remove()    
used to remove or delete a file. this method can not remove or delete a directory.If specified path is directory then OSError will be raised by the method.
- os.rmdir()    
This method is used to remove or delete an empty directory.OSError will be raised if the specified path is not an empty directory

```python
%%writefile stringfun.py
def stringfunctions(x):
    print('Upper:',x.upper())
    print('Lower:',x.lower())
    print('Length:',len(x))
    print('Capitalize:',x.capitalize())
    print('Title:',x.title())
    print('Digit:',x.isdigit())
    print('Is upper:',x.isupper())
    print('Is Lower:',x.islower())
    print('Alpha digit:',x.isalnum())
    print('Alpha:',x.isalpha())
    print(x)
```
