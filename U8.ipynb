### try except else finally
- The try block lets you test a block of code for errors.
- The except block lets you handle the error.
- The else block lets you execute the code when there is not error.
- The finally block lets you execute the code regardless of the result if the try and except block.

```python
def div(a,b):
    return a/b
try:
    print(div(5,0))
except:
    print('Exception occured')
else:
    print('No exception')
finally:
    print('This always prints')
```

```python
# file not found Error
# name Error
# zero division Error
# Index Error
try:
    m=5
    f=open('abc.txt','r')
    print(f.read())
    print(m)
    print(5/2)
    l=[1,2,3]
    print(l[100])
except FileNotFoundError:
    print('File not found')
except NameError:
    print('Variable not defined')
except ZeroDivisionError:
    print("Can't divide by Zero")
except IndexError:
    print('Index out of range')
```
