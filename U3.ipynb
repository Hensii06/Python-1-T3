### Task 9 - write a program to read line by line from the given files File 1 and File 2 and write into file 3

```python
f1=open('Task 9.1.txt')
f2=open('Task 9.2.txt')
x=f1.readlines()
y=f2.readlines()
flag=True
f3=open('Task 9.3.txt','w+')
if len(x)>len(y):
    l=len(y)
else:
    l=len(x)
    flag=False
for i in range(l):
    f3.write(x[i])
    if not x[i].endswith('\n'):
            f3.write('\n')
    f3.write(y[i])
    if not y[i].endswith('\n'):
            f3.write('\n')
if flag:
    u=x[l:]
    f3.writelines(u)
else:
    u=y[l:]
    f3.writelines(u)
f1.close()
f2.close()
f3.close()
```

### Task 11 - write a program to reverse content of one file and store it in second file and also convert a content of second file into uppercase and store it in 3rd file and also count number of vowels in 3rd file and print only second line from the content of 3rd file

```python
with open('Task 11.txt') as f1:
    x=f1.read()
    with open('Task 11_1.txt','w') as rev:
        rev.write(x[::-1])
with open('Task 11_1.txt','r') as rev:
    content=rev.read().upper()
    with open('Task 11_2.txt','w') as upper:
        upper.write(content)
count=0
with open('Task 11_2.txt','r') as upper:
    content=upper.read()
    for i in content:
        if i in 'aeiouAEIOU':
            count=count+1
    upper.seek(0)
    x1=upper.readlines()[1]
print(x1)
print('Total vowels:',count)
```

### Task 12 - write a program to read a text file and reverse the word starting with i

```python
f=open('Task 12.txt')
x=f.read().split()
print(x)
new=''
for i in x:
    if i[0]=='I' or i[0]=='i':
        new=new+i[::-1]+' '
    else:
        new=new+i+' '
print(new)
f.close()
```
