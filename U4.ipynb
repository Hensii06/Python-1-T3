### Task 14 (PB 486) - Write a python program to accept string/sentence from user till the user enters “END”. Each string/sentence entered by user should be a newline in file. Save all the lines in file and display only those lines which begin with capital letter.

```python
f=open('Task 14.txt','w+')
while True:
    u=input("Enter sentances (Enter 'END' to quit):")
    if u=='END':
        break
    else:
        f.write(u+'\n')
f.seek(0)
for line in f:
    if line[0].isupper():
        print(line)
f.close()
```

### Task 15 - write a program to remove blank line from a file

```python
with open('Task 15.txt', 'r') as f:
    lines = f.readlines()
with open('Task 15.txt', 'w') as f:
    for line in lines:
        if line.strip():  
            f.write(line)
            print(line)
```

### Task 16 - write a program to find most frequntly occuring word from the given text file.

```python
with open('Task 16.txt') as f:
    words = f.read().lower().split()
fr={}
for word in words:
    word=word.strip('.,?!')
    if word in fr:
        fr[word]=fr[word]+1
    else:
        fr[word]=1
x = max(fr, key=fr.get)
y = fr[x]
print('most freq word is',x)
print('word count is',y)
```

### Task 17 - write a program to store employee details in text file and calculate the avg salary

```python
n=int(input('Enter number of employees:'))
with open('Task 17.txt','w') as f:
    for i in range(n):
        name=input('Enter employee name:')
        salary=float(input('Enter salary:'))
        f.write(name+','+str(salary)+'\n')
total=0
count=0
with open('Task 17.txt') as f:
    for line in f:
        name,salary=line.strip().split(',')
        total=total+float(salary)
        count=count+1
if count>0:
    print('avg salary:',total/count)
```
