## Writing the files
- write() - write the content in the file without adding any extra characters
- writelines() - write the content of a list into file

```python
f=open('H2W.txt','w')
f.write('This is Demo File\n')
f.write('This is Demo Text')
f.close()
```

```python
l=['This is demo for writelines()\n','This is demo text\n','LJ University\n']
f=open('H2w.txt','w+')
f.writelines(l)
f.seek(0)
print(f.read())
f.close()
```

```python
f=open('name.txt','w+')
for i in range(5):
    u=input("enter name")
    f.write(u+'\n')
f.seek(0)
print(f.read())
f.close()
```

### Task 5 - write a program to copy content of file but reverse content

```python
f1=open('Task 5.txt','r')
f2=open('Task 5_1.txt','w')
lines=[]
line=f1.readline()
while line!='':
    if not line.endswith('\n'):
        line=line+'\n'
    lines.append(line)
    line=f1.readline()
lines.reverse()
for l in lines:
    f2.write(l)
f1.close()
f2.close()
```

### Task 7 - write a program to count words,characters and spaces from the text file

```python
f=open('Task 7.txt')
x=f.read()
y=x.split()
print('Word count :',len(y))
print('space count :',x.count(' '))
print('Character count :',len(x)-x.count(' ')-x.count('\n'))
f.close()
```
